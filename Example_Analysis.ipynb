## Now investigate particular examples from the subtasks listed above
Given the values found above, now find particular examples to see if any information can be found from it 

```python
from analysis import get_docs
from random import randint
```

```python
def get_doc_one_right_doc_two_wrong(model_one_docs, model_two_docs):
    filtered_docs = []
    for idx in range(len(model_one_docs)):
        if model_one_docs[idx]["acc"]==1.0 and model_two_docs[idx]["acc"]==0.0:
            filtered_docs.append(model_one_docs[idx])
    return filtered_docs
```

```python
model_one_docs = get_docs.get_docs(model_one, subtask1, {"ContextType":"direct"})
model_two_docs = get_docs.get_docs(model_two, subtask1, {"ContextType":"direct"})
assert len(model_one_docs)==len(model_two_docs)
material_properties_filtered = get_doc_one_right_doc_two_wrong(model_one_docs, model_two_docs)
```

```python
random_doc = material_properties_filtered[randint(0, len(material_properties_filtered)-1)]
```

```python
context = random_doc["Context1"]
target = random_doc["Target1"]
```

### Visualization


```python
from analysis import tree_functions
from transformers import AutoTokenizer, AutoModelForMaskedLM, AutoConfig, utils
from bertviz import model_view
```

```python
utils.logging.set_verbosity_error()  # Suppress standard warnings
# model_two 
baseline_bert_model = AutoModelForMaskedLM.from_pretrained("babylm/ltgbert-10m-2024", trust_remote_code=True, return_dict=True, output_attentions=True)
baseline_bert_tokenizer = AutoTokenizer.from_pretrained("babylm/ltgbert-10m-2024", trust_remote_code=True)
parser_bert = AutoModelForMaskedLM.from_pretrained("/work/vesuvius/babylm/models/ELC_PaserBERT_10M_256_batch", tree_output=True, trust_remote_code=True, output_attentions=True)
parser_tokenizer = AutoTokenizer.from_pretrained("/work/vesuvius/babylm/models/ELC_PaserBERT_10M_256_batch", trust_remote_code=True)
```

```python
totality = f"{context} {target}"
tree = tree_functions.model_to_tree_output(totality, parser_tokenizer, parser_bert)
```

```python
# elc-bert visualization
elc_bert_inputs = baseline_bert_tokenizer.encode(totality, return_tensors='pt')  # Tokenize input text
baseline_outputs = baseline_bert_model(elc_bert_inputs, output_attentions=True)  # Run model
baseline_attention = baseline_outputs[-1]  # Retrieve attention from model outputs
baseline_tokens = baseline_bert_tokenizer.convert_ids_to_tokens(elc_bert_inputs[0])  # Convert input ids to token strings
model_view(baseline_attention, baseline_tokens)
```

```python
parser_bert_inputs = parser_tokenizer.encode_plus(context, target, return_tensors='pt')  # Tokenize input text
parser_outputs = parser_bert(parser_bert_inputs, output_attentions=True)  # Run model
parser_attention = parser_outputs[-1]  # Retrieve attention from model outputs
parser_tokens = parser_tokenizer.convert_ids_to_tokens(parser_bert_inputs[0])  # Convert input ids to token strings
model_view(parser_attention, parser_tokens)
```
